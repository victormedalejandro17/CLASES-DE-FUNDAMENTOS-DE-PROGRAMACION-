# Clase 9: NumPy, Pandas y gráficas
**Objetivo:** usar arreglos, tablas, operaciones estadísticas, filtros, agrupaciones y gráficas con un ejemplo distinto.

**Caso:** seguimiento de ventas y devoluciones de una pequeña tienda.

## 1. Importar bibliotecas

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Bibliotecas listas")
```

## 2. NumPy: arreglos y operaciones

```python
# Ventas diarias (en cientos de pesos)
ventas = np.array([12, 15, 11, 18, 14, 20])

print("Ventas:", ventas)
print("Forma:", ventas.shape)
print("Total:", ventas.sum())
print("Promedio:", ventas.mean())
print("Mínimo:", ventas.min())
print("Máximo:", ventas.max())

# Convertir a pesos
pesos = ventas * 100
print("Ventas en pesos:", pesos)
```

## 3. Comparar y filtrar valores

```python
# Identificar días con ventas mayores a 14
print(ventas > 14)
print("Días con ventas altas:", ventas[ventas > 14])
```

## 4. Crear una tabla con Pandas

```python
datos = {
    "Dia": [1, 2, 3, 4, 5, 6],
    "Sucursal": ["Centro", "Norte", "Centro", "Norte", "Centro", "Norte"],
    "Ventas": [1200, 1500, 1100, 1800, 1400, 2000],
    "Devoluciones": [2, 3, 1, 4, 2, 3]
}

df = pd.DataFrame(datos)
df
```

## 5. Agregar columnas calculadas

```python
df["Venta_neta"] = df["Ventas"] - df["Devoluciones"] * 50
df["Porcentaje_devolucion"] = (
    df["Devoluciones"] / (df["Ventas"] / 100)
).round(2)

df
```

## 6. Seleccionar, filtrar y ordenar

```python
# Seleccionar columnas
display(df[["Sucursal", "Ventas"]])

# Filtrar ventas mayores o iguales a 1500
display(df[df["Ventas"] >= 1500])

# Ordenar de mayor a menor venta
display(df.sort_values("Ventas", ascending=False))
```

## 7. Agrupar y resumir

```python
resumen = df.groupby("Sucursal")[["Ventas", "Devoluciones"]].sum()
resumen["Porcentaje_devolucion"] = (
    resumen["Devoluciones"] / (resumen["Ventas"] / 100)
).round(2)

resumen
```

## 8. Gráfica de barras

```python
resumen["Ventas"].plot(kind="bar", title="Ventas por sucursal",
                      ylabel="Pesos", xlabel="Sucursal", rot=0)
plt.tight_layout()
plt.show()
```

## 9. Gráfica de líneas

```python
df.pivot(index="Dia", columns="Sucursal", values="Ventas").plot(
    marker="o", title="Ventas por día", ylabel="Pesos"
)
plt.tight_layout()
plt.show()
```

## 10. Gráfica de dispersión

```python
plt.scatter(df["Ventas"], df["Devoluciones"])
plt.title("Ventas y devoluciones")
plt.xlabel("Ventas (pesos)")
plt.ylabel("Número de devoluciones")
plt.tight_layout()
plt.show()
```

### Conclusión
Con NumPy calculamos estadísticas y filtramos arreglos. Con Pandas organizamos, calculamos y resumimos datos. Matplotlib permite visualizar los resultados.
