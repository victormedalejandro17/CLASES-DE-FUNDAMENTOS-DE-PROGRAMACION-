## Índices anidados

### Tuplas e índices anidados

```
tupla = (1, [4, True, None, 'Saludo'])
```

La siguiente celda regresa la lista que se encuentra en el índice 1 de la tupla.

```
tupla[1]
```

La siguiente celda regresa el objeto `'Saludo'` a partir de `tupla[1][3]`.

```
tupla[1][3]
```

```
tupla[1][3][2]
```

## Diccionarios

A continuación se define el objeto de tipo `dict` con el nombre `persona`.

```
persona = {'nombre': 'Juan', 'primer apellido': 'Pérez', 'promedio': 7.8, True: 'V'}
```

La siguiente celda obtiene el objeto `7.8` correspondiente a la clave `'promedio'` del objeto persona.

```
persona['promedio']
```

La siguiente celda obtiene el objeto `'Juan'` correspondiente a la clave `'nombre'` del objeto persona.

```
persona['nombre']
```

La siguiente celda obtiene el objeto `'V'` correspondiente a la clave `True` del objeto persona.

```
persona[True]
```

La siguiente celda hace referencia a la clave `'indefinido'`, la cual no existe en el objeto persona, por lo que genera un error de tipo `KeyError`.

```
persona['indefinido']
```

## Indexado de elementos por rangos numéricos

Las colecciones con índices numéricos como `tuple`, `list`, `str`, `bytearray` y `bytes` permiten indexar rangos de elementos con la siguiente sintaxis:

`<colección>[<m>:<n>:<i>]`

- `m` es el índice inicial.
- `n` es el índice final.
- `i` indica los incrementos (si es positivo) o decrementos (si es negativo).

El resultado será un nuevo objeto del mismo tipo que el original, incluyendo los elementos comprendidos entre el índice inicial y uno antes del índice final, según los incrementos o decrementos indicados. A esto se le conoce como **rebanado**.

### Ejemplos

Para ilustrar el uso del rebanado se crea el objeto `lista` de tipo `list`.

```
lista = [0, 1, 2, 3, 4, 5, 6, 7, 8]
```

La siguiente celda regresa los elementos con índices que van de 6 a antes de 1 en decrementos de 1.

```
lista[6:1:-1]
```

La siguiente celda regresa los elementos con índices que van de 6 a antes de 0 en decrementos de 2.

```
lista[6:0:-2]
```

La siguiente celda regresa una lista vacía, en vista de que el rango no tiene sentido.

```
lista[0:6:-2]
```

La siguiente celda regresa un objeto tipo `list` con los elementos con índices 2 a 4.

```
lista[2:5]
```
