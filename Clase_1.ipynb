# Clase 1

## Introducción a Python

En esta clase vamos a ver cómo guardar información en variables y cómo mostrarla en pantalla usando `print()`.

## El espacio de nombres (namespace)

El espacio de nombres guarda la relación entre los nombres que usamos y los objetos que existen en Python. Por ejemplo, cuando creamos una variable, le damos un nombre a un valor para poder utilizarlo después.

## El operador de asignación (=)

El signo `=` sirve para guardar un valor dentro de una variable. Primero se escribe el nombre y después el valor que queremos guardar.

```
saludo = "Hola, ¿cómo estás?"
print(saludo)
```

## Ejemplo con números

También podemos guardar números enteros y decimales en variables. Después podemos hacer operaciones con ellos.

```
numero = 8
print(numero)

numero2 = numero + 2
print(numero2)

numero3 = 5
print(numero3)

resultado = numero2 + numero3
print(resultado)

precio = 24.50
print(precio)
```

## Valores booleanos

Los valores `True` y `False` sirven para representar si algo es verdadero o falso.

```
tarea_entregada = True
examen_aprobado = False

print(tarea_entregada)
print(examen_aprobado)
```

## Para recordar

- Una variable sirve para guardar información.
- El signo `=` asigna un valor.
- `print()` muestra el resultado en pantalla.
- `True` y `False` representan valores verdaderos o falsos.
