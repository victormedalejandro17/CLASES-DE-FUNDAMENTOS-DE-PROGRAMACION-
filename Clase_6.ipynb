# S6 — Conversiones y operadores en Python

**Ejercicios de práctica**

Este cuaderno resume los procesos de la clase: consultar tipos, convertir datos y aplicar operadores aritméticos y de colecciones. Los ejemplos usan datos distintos y están preparados para ejecutarse sin errores.

## 1. Función `type()`

Indica el tipo de dato de un valor.

```python
valores = ["Tampico", 18, 4.5, 2+3j, True, [1, 2], (3, 4), {5, 6}, {"curso": "Python"}]
for valor in valores:
    print(valor, "->", type(valor).__name__)
```

## 2. Conversiones numéricas

`int()` convierte a entero (trunca decimales); `float()` convierte a decimal; `complex()` crea un número complejo. `bool()` evalúa si un valor es verdadero o falso.

```python
print("int:", int("24"), int(7.9), int(False))
print("float:", float("6.25"), float(8), float(True))
print("complex:", complex(4, 2), complex("3+5j"), complex(9))
print("bool:", bool(0), bool(""), bool([1, 2]), bool(None))
```

## 3. Conversión a texto y bytes

`str()` produce una cadena. `bytes()` codifica texto con una codificación indicada o crea bytes de una longitud.

```python
precio = 39.5
texto = str(precio)
contenido = "Hola Python".encode("utf-8")
print(texto, type(texto).__name__)
print(contenido, type(contenido).__name__)
print(contenido.decode("utf-8"))
print(bytes(4))  # cuatro bytes con valor cero
```

## 4. Conversión controlada

Antes de convertir una entrada, podemos comprobar que tenga el formato esperado. Así evitamos que una entrada inválida detenga el programa.

```python
entrada = "125"
if entrada.isdigit():
    numero = int(entrada)
    print("Conversión correcta:", numero)
else:
    print("Escribe un número entero válido.")
```

## 5. Operadores aritméticos

`+` suma, `-` resta, `*` multiplica, `**` potencia, `/` división real, `//` división entera y `%` residuo.

```python
a, b = 17, 5
print("Suma:", a + b)
print("Resta:", a - b)
print("Producto:", a * b)
print("Potencia:", b ** 2)
print("División:", a / b)
print("División entera:", a // b)
print("Residuo:", a % b)
```

## 6. Precedencia y paréntesis

Primero se resuelven los paréntesis, luego las potencias, multiplicaciones/divisiones y finalmente sumas/restas. Los paréntesis permiten cambiar el orden.

```python
resultado_1 = 6 + 3 * 4
resultado_2 = (6 + 3) * 4
resultado_3 = 2 ** 3 + 10 / 2
print(resultado_1, resultado_2, resultado_3)
```

## 7. Operadores en colecciones

El operador `+` concatena colecciones compatibles del mismo tipo.

```python
nombre = "Ana"
apellido = " López"
print("Cadena:", nombre + apellido)
print("Lista:", [10, 20] + [30, 40])
print("Tupla:", ("rojo", "azul") + ("verde",))
print("Bytes:", b"Py" + b"thon")
```

## 8. Ejercicio integrador

Calcula el total de una compra, convierte el total a texto y muestra el tipo de dato.

```python
cantidad = 3
costo_unitario = 12.50
total = cantidad * costo_unitario
resumen = "Total de la compra: $" + str(total)
print(resumen)
print("Tipo del total:", type(total).__name__)
print("¿Total positivo?", bool(total))
```
