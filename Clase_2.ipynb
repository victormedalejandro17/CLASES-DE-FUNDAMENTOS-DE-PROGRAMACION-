# Clase 2

## Expresiones y comentarios

En esta clase veremos algunos ejemplos de expresiones en Python y cómo usar comentarios para explicar el código.

## Expresiones

Una expresión combina valores, variables u operadores y al ejecutarse produce un resultado.

**Ejemplo:** guardamos un número en una variable y después lo mostramos.

```
numero = 25.5
numero
```

## Expresiones aritméticas

Se utilizan para realizar operaciones matemáticas, como sumar, restar, multiplicar o dividir.

```
12 + 8
```

```
20 - 6
```

```
4 * 5
```

```
18 / 3
```

## Expresiones de relación

Comparan dos valores. El resultado será `True` si la condición se cumple o `False` si no se cumple.

```
15 >= 10
```

```
7 == 9
```

## Métodos y funciones

El método `upper()` convierte un texto a letras mayúsculas. La función `len()` cuenta cuántos caracteres tiene un texto.

```
"universidad".upper()
```

```
len('Python')
```

## Varias expresiones en una sola línea

Se pueden escribir varias instrucciones en una línea separándolas con punto y coma (`;`). Python las ejecuta en orden.

```
a = 10; b = 5; a + b
```

## Comentarios de una sola línea

Los comentarios empiezan con el símbolo `#`. Sirven para explicar el código y Python no los ejecuta.

```
# Guardamos el nombre y lo mostramos
nombre = 'Carlos'
print(nombre)
```

## Docstrings

Los docstrings son textos que se escriben entre triples comillas. Se usan para explicar qué hace una función.

```
def saludo():
    """Muestra un saludo en pantalla."""
    print("Hola, bienvenido")

saludo()
```

## Para recordar

- Una expresión produce un resultado.
- Los operadores aritméticos realizan cálculos.
- Las expresiones de relación comparan valores.
- `upper()` cambia un texto a mayúsculas y `len()` cuenta sus caracteres.
- `#` permite escribir comentarios.
- Los docstrings explican la función de un bloque de código.
