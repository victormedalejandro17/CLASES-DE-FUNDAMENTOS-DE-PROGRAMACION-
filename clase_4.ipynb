# Clase 4: Diccionarios, conjuntos y conjuntos inmutables

En este ejemplo se muestran ejercicios parecidos a los de la clase, pero con datos diferentes.
Cada apartado incluye el código y una breve explicación del resultado.

## 1. Tipo `dict`
Un diccionario almacena información en pares de **clave: valor**.

```python
# Crear un diccionario con datos de un estudiante
estudiante = {
    'nombre': 'Mariana',
    'apellido': 'Lopez',
    'semestre': 2,
    'promedio': 8.6
}

estudiante
```

## 2. Diccionario vacío
Se puede crear un diccionario vacío usando `{}`.

```python
datos = {}
datos
```

## 3. Claves repetidas en un diccionario
Si una clave aparece más de una vez, se conserva el último valor asignado.

```python
producto = {
    'nombre': 'Cuaderno',
    'precio': 35,
    'nombre': 'Agenda'
}

producto
```

## 4. Error al usar una lista como clave
Las claves de un diccionario deben ser inmutables. Una lista es mutable, por eso produce `TypeError`.

```python
try:
    ejemplo = {['rojo', 'azul']: 'colores'}
except TypeError as error:
    print(type(error).__name__ + ':', error)
```

## 5. Tipo `set`
Un conjunto almacena elementos únicos. Si hay valores repetidos, sólo aparece uno.

```python
numeros = {4, 2, 4, 7, 2, 9, 4}
numeros
```

## 6. Valores equivalentes dentro de un `set`
En Python, `1` y `True` se consideran equivalentes al comparar elementos de un conjunto; por eso sólo se conserva uno.

```python
valores = {1, 2, True, False, 0, 3}
valores
```

## 7. Error al incluir una lista en un `set`
Los elementos de un conjunto deben ser inmutables. Una lista no puede agregarse directamente.

```python
try:
    conjunto = {'lunes', ['martes', 'miércoles'], 'jueves'}
except TypeError as error:
    print(type(error).__name__ + ':', error)
```

## 8. Error al incluir un diccionario en un `set`
Un diccionario también es mutable, por lo que no puede ser un elemento de un conjunto.

```python
try:
    conjunto = {'dato', {'curso': 'Python'}}
except TypeError as error:
    print(type(error).__name__ + ':', error)
```

## 9. Tipo `frozenset`
Un `frozenset` es un conjunto inmutable. Se crea con la función `frozenset()` y recibe una colección como argumento.

```python
frozenset('Tampico')
```

## 10. Crear un `frozenset` desde una lista

```python
frozenset([5, 8, 5, 2, 8])
```

## 11. Crear un `frozenset` desde una tupla

```python
frozenset(('norte', 'sur', 'este'))
```

## 12. Crear un `frozenset` desde un diccionario
Cuando recibe un diccionario, el `frozenset` toma sus claves.

```python
datos = {'materia': 'Programación', 'grupo': 'A', 'turno': 'matutino'}
frozenset(datos)
```

## 13. `frozenset` vacío

```python
frozenset()
```

## 14. Error al incluir una lista dentro de un `frozenset`
Aunque `frozenset` es inmutable, sus elementos también deben ser inmutables. Una lista provoca `TypeError`.

```python
try:
    frozenset((10, ['a', 'b'], 20))
except TypeError as error:
    print(type(error).__name__ + ':', error)
```

## 15. Error al pasar varios argumentos
`frozenset()` acepta como máximo un argumento: la colección que se desea convertir.

```python
try:
    frozenset('sol', 'luna', 'mar')
except TypeError as error:
    print(type(error).__name__ + ':', error)
```

## Conclusión
- `dict`: guarda pares de clave y valor.
- `set`: guarda elementos únicos y permite modificarlos.
- `frozenset`: guarda elementos únicos y no permite modificarlos.
- Las listas y los diccionarios no pueden usarse como elementos de `set` o como claves de `dict` porque son mutables.
