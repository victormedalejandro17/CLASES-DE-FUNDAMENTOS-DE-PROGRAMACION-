# Clase 8
## Ciclos `for`, función `range()` y desempaquetado

**Objetivo:** practicar recorridos numéricos y de colecciones, y asignar elementos a varias variables.

### 1. La función `range()`

`range(inicio, límite, paso)` genera una secuencia. El límite no se incluye; el paso puede ser positivo o negativo.

```python
# Contar del 1 al 5
for numero in range(1, 6):
    print(numero)
```

```python
# Avanzar de 2 en 2
for numero in range(2, 11, 2):
    print(numero)
```

```python
# Contar hacia atrás
for numero in range(10, 0, -2):
    print(numero)
```

### 2. Desempaquetado de variables

Se pueden asignar los elementos de una colección a varias variables. La cantidad de variables debe coincidir con la cantidad de elementos.

```python
producto, precio, unidades = ["Cuaderno", 28.5, 3]

print(producto)
print(precio)
print(unidades)
```

```python
# También se puede desempaquetar una tupla
x, y, z = (4, 7, 9)
print(x, y, z)
```

### 3. Desempaquetado con `for`

En cada vuelta del ciclo, Python puede asignar los elementos de cada subcolección a variables distintas.

```python
pares = [("Ana", 8), ("Luis", 9), ("Mia", 10)]

for nombre, calificacion in pares:
    print(nombre, "obtuvo", calificacion)
```

```python
# Recorrer palabras y mostrar sus letras
palabras = ["sol", "luz", "mar"]

for palabra in palabras:
    primera, segunda, tercera = palabra
    print(primera, segunda, tercera)
```

### 4. Práctica breve

Modifica los ejemplos para contar de 3 a 15 de 3 en 3 y para mostrar el nombre y la edad de dos personas.

```python
# Solución de práctica: contar de 3 a 15
for numero in range(3, 16, 3):
    print(numero)

personas = [("Eva", 19), ("Diego", 21)]
for nombre, edad in personas:
    print(f"{nombre}: {edad} años")
```
