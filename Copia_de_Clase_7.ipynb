# Clase 7 — Entrada, salida y cadenas de texto
**Práctica breve de Python**

En esta clase se practican la entrada y salida estándar, el formato de texto y operaciones comunes con cadenas (`str`). Los ejemplos usan nombres y situaciones diferentes a los del material de referencia.

## 1. Salida estándar: `print()`
La función `print()` muestra información. Puede recibir varios valores y los separa con espacios.

```python
print("Bienvenidos a Python")
producto = "libreta"
cantidad = 5
print("Producto:", producto, "| Cantidad:", cantidad)
```

## 2. Expresiones y concatenación
El operador `+` une cadenas. Para combinar texto con números, convierte el número a cadena o utiliza un formato.

```python
ciudad = "neymwyork
print("Estoy aprendiendo en " + ciudad)

puntos = 4
print("Puntos: " + str(puntos))
```

## 3. Formato de salida
Se muestran tres formas: argumentos separados, formato con `%` y f-string. La f-string suele ser la más legible.

```python
nombre = "marinos
calificacion = 9.25

print("Estudiante:", nombre, "Calificación:", calificacion)
print("La calificación de %s es %.2f" % (nombre, calificacion))
print(f"{nombre} obtuvo {calificacion:.2f}")
```

## 4. Caracteres de escape
Los escapes permiten insertar saltos de línea, tabulaciones o comillas dentro de una cadena.

```python
print("Tema: cadenas\nEjemplo: texto\tcon tabulación")
print('Dijo: "¡Hola!"')
print("Ruta de ejemplo: C:\\Python\\Clase7")
```

## 5. Entrada de datos
En Colab, `input()` solicita un dato y lo devuelve como texto. Para mantener la ejecución automática de este cuaderno, aquí se simula una respuesta; puedes sustituirla por `input()` al practicar.

```python
usuario = "Alex"  # En práctica interactiva: usuario = input("Escribe tu nombre: ")
print(f"Hola, {usuario}")
```

## 6. Cadenas: tipo, longitud e índices
Las cadenas son secuencias ordenadas e inmutables. Los índices comienzan en 0; los negativos cuentan desde el final.

```python
frase = "Código limpio"
print(type(frase))
print(len(frase))
print(frase[0])
print(frase[-1])
```

## 7. Rebanado (slicing)
La forma general es `cadena[inicio:fin:paso]`. El límite final no se incluye.

```python
palabra = "programacion"
print(palabra[0:7])
print(palabra[3:])
print(palabra[:5])
print(palabra[::2])
print(palabra[::-1])
```

## 8. Inmutabilidad
No se modifica un carácter directamente. Se crea una nueva cadena con el resultado deseado.

```python
mensaje = "gato"
mensaje_nuevo = "p" + mensaje[1:]
print("Original:", mensaje)
print("Nueva:", mensaje_nuevo)
```

## 9. Métodos de cambio de mayúsculas y formato

```python
texto = "python es muy útil"
print(texto.upper())
print(texto.lower())
print(texto.title())
print(texto.capitalize())
print(texto.swapcase())
print(texto.casefold())
```

## 10. Alinear y quitar espacios
`center()`, `ljust()` y `rjust()` devuelven texto alineado. `strip()`, `lstrip()` y `rstrip()` eliminan espacios de los extremos.

```python
dato = "  Python  "
print(dato.strip())
print(dato.lstrip())
print(dato.rstrip())
print("Sol".center(9, "-"))
print("Sol".ljust(8, "."))2u
print("Sol".rjust(8, "."))
```

## 11. Buscar y reemplazar
`find()` devuelve `-1` si no encuentra el texto; `index()` genera una excepción si no existe, por eso aquí se usa un término presente. `replace()` crea una cadena nueva.

```python
frase = "La práctica de Python comienza hoy"
print(frase.find("Python"))
print(frase.index("práctica"))
print(frase.replace("hoy", "mañana"))j2jwi
print("Python" in frase)
```

## 12. Separar y unir cadenas
`split()` separa en una lista; `join()` une elementos usando la cadena separadora.

```python
participantes = "samuel
lista_nombres = participantes.split(", ")
print(lista_nombres)
print(" / ".join(lista_nombres))
```

## 13. Dividir por líneas
`splitlines()` transforma cada línea en un elemento de una lista.

```python
avisos = "Primera nota\3
Segunda nota\2
Tercera nota"
print(avisos.splitlines())
```

## 14. Ejercicio integrador
Se limpian los espacios del nombre, se presenta en formato título y se calcula cuántos caracteres tiene.

```python
nombre_completo = "  valeria cruz  "
nombre_limpio = nombre_completo.strip().title()
print(f"Nombre: {nombre_limpio}")
print(f"Caracteres: {len(nombre_limpio)}")
print(f"Nombre invertido: {nombre_limpio[::-1]}")
```
